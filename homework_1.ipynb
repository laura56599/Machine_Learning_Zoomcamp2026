## Q6. Mediana de `horsepower` antes y después de imputar con la moda

```python
mediana_antes = df['horsepower'].median()
moda = df['horsepower'].mode()[0]
mediana_despues = df['horsepower'].fillna(moda).median()
print('Mediana antes:', mediana_antes)
print('Moda:', moda)
print('Mediana después:', mediana_despues)
```

## Q7. Suma de pesos (regresión lineal con la ecuación normal)
$w = (X^TX)^{-1}X^Ty$

```python
X = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7).values
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
print(w)
w.sum()
```
